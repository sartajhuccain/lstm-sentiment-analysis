```python
import os
import re
import json
import random
import pickle
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Embedding,
    Bidirectional,
    LSTM,
    Dense,
    Dropout,
    SpatialDropout1D
)
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
print("GPU available:", len(tf.config.list_physical_devices("GPU")) > 0)
```

```python
from urllib.parse import quote

RECORD_ID = "7928582"
API_URL = f"https://zenodo.org/api/records/{RECORD_ID}"

response = requests.get(API_URL, timeout=60)
response.raise_for_status()

record = response.json()

csv_files = [
    f for f in record["files"]
    if f["key"].lower().endswith(".csv")
]

if not csv_files:
    raise FileNotFoundError(
        "CSV file nahi mili. Official dataset page check karein."
    )

file_info = next(
    (f for f in csv_files
     if f["key"].lower() == "imdb dataset.csv"),
    csv_files[0]
)

download_url = file_info["links"]["self"]
CSV_PATH = "/content/IMDB Dataset.csv"

with requests.get(
    download_url, stream=True, timeout=(30, 180)
) as r:
    r.raise_for_status()
    with open(CSV_PATH, "wb") as out:
        for chunk in r.iter_content(chunk_size=1024 * 1024):
            if chunk:
                out.write(chunk)

print("Dataset downloaded:", CSV_PATH)
print("File size (MB):", round(os.path.getsize(CSV_PATH) / 1e6, 2))
```

```python
df = pd.read_csv(CSV_PATH)

# Column names ko standardize karein
df.columns = df.columns.str.strip().str.lower()

print("Dataset shape:", df.shape)
print("\nColumns:", df.columns.tolist())
display(df.head())

print("\nMissing values:")
print(df.isnull().sum())

print("\nSentiment distribution:")
print(df["sentiment"].value_counts())
```

```python
df = df[["review", "sentiment"]].dropna().copy()

df["review"] = df["review"].astype(str)
df["sentiment"] = (
    df["sentiment"].astype(str).str.strip().str.lower()
)

# HTML tags remove karein
df["review"] = df["review"].apply(
    lambda x: re.sub(r"<[^>]+>", " ", x)
)

# Extra spaces remove karein
df["review"] = df["review"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)

# Empty reviews aur unexpected labels remove karein
df = df[df["review"].str.len() > 0]
df = df[df["sentiment"].isin(["positive", "negative"])]

# Duplicate records remove karein
df = df.drop_duplicates(subset=["review"]).reset_index(drop=True)

# Positive = 1, Negative = 0
df["label"] = df["sentiment"].map({
    "negative": 0,
    "positive": 1
}).astype("int32")

print("Clean dataset shape:", df.shape)
display(df.head())
```

```python
counts = df["sentiment"].value_counts()

plt.figure(figsize=(7, 4))
counts.plot(kind="bar")
plt.title("IMDb Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
X = df["review"].values
y = df["label"].values

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
MAX_WORDS = 30000
MAX_LEN = 250
OOV_TOKEN = "<OOV>"

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token=OOV_TOKEN
)

# Sirf training text par vocabulary fit karein
tokenizer.fit_on_texts(X_train)

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

print("Training input shape:", X_train_pad.shape)
print("Testing input shape:", X_test_pad.shape)
print("Vocabulary size:", min(
    MAX_WORDS, len(tokenizer.word_index) + 1
))
```

```python
VOCAB_SIZE = min(
    MAX_WORDS, len(tokenizer.word_index) + 1
)

model = Sequential([
    Input(shape=(MAX_LEN,)),

    Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=128,
        mask_zero=True
    ),

    SpatialDropout1D(0.25),

    Bidirectional(
        LSTM(64, dropout=0.2)
    ),

    Dense(64, activation="relu"),
    Dropout(0.4),

    Dense(1, activation="sigmoid")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=1,
    min_lr=1e-6,
    verbose=1
)

history = model.fit(
    X_train_pad,
    y_train,
    validation_split=0.15,
    epochs=8,
    batch_size=128,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(history.history["accuracy"], label="Train")
axes[0].plot(history.history["val_accuracy"], label="Validation")
axes[0].set_title("Model Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(history.history["loss"], label="Train")
axes[1].plot(history.history["val_loss"], label="Validation")
axes[1].set_title("Model Loss")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Loss")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
y_prob = model.predict(X_test_pad, batch_size=128).ravel()
y_pred = (y_prob >= 0.5).astype(int)

print("Test Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall:", round(recall_score(y_test, y_pred), 4))
print("F1 Score:", round(f1_score(y_test, y_pred), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Negative", "Positive"],
    digits=4
))
```

```python
def predict_sentiment(review_text):
    cleaned = re.sub(r"<[^>]+>", " ", str(review_text))
    cleaned = re.sub(r"\s+", " ", cleaned).strip()

    sequence = tokenizer.texts_to_sequences([cleaned])
    padded = pad_sequences(
        sequence,
        maxlen=MAX_LEN,
        padding="post",
        truncating="post"
    )

    probability = float(
        model.predict(padded, verbose=0)[0][0]
    )

    sentiment = "Positive" if probability >= 0.5 else "Negative"
    confidence = probability if probability >= 0.5 else 1 - probability

    print("Review:", review_text)
    print("Predicted sentiment:", sentiment)
    print(f"Model confidence score: {confidence:.2%}")

    return sentiment

predict_sentiment(
    "This movie was amazing, the acting was brilliant and the story was beautiful."
)

predict_sentiment(
    "The movie was boring, poorly acted and a complete waste of time."
)
```
